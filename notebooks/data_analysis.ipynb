```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("../data/disease_data.csv")

# Display the first five rows
df.head()
```

```python
# Dataset Information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Summary:")
df.describe()
```

```python
# Disease Outbreak Class Distribution

import matplotlib.pyplot as plt

# Count the records in each outbreak class
outbreak_counts = df["outbreak"].value_counts().sort_index()

print("Outbreak Class Distribution:")
print(outbreak_counts)

# Create a bar chart
outbreak_counts.plot(
    kind="bar",
    title="Disease Outbreak Class Distribution"
)

plt.xlabel("Outbreak Class")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Feature Analysis

print("Average Temperature:", df["temperature"].mean())
print("Average Rainfall:", df["rainfall"].mean())
print("Average Reported Cases:", df["cases"].mean())

print("\nFeature Correlation:")
print(df[["temperature", "rainfall", "cases", "outbreak"]].corr())
```

```python
# Feature Analysis with Graphs

import matplotlib.pyplot as plt

# Calculate average values
averages = {
    "Temperature (°C)": df["temperature"].mean(),
    "Rainfall (mm)": df["rainfall"].mean(),
    "Reported Cases": df["cases"].mean()
}

# Display average values
for feature, value in averages.items():
    print(f"{feature}: {value:.2f}")

# Create a bar graph
plt.figure(figsize=(9, 5))

plt.bar(averages.keys(), averages.values())

plt.title("Average Disease-Related Features")
plt.xlabel("Features")
plt.ylabel("Average Value")

plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```
